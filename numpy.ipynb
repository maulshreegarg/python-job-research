```python
#df_usa.drop(index="job_index", inplace=True)
```

```python
df_usa=df_usa.reset_index()
```

```python
df_usa.set_index("job_index",inplace=True)
```

```python
df_usa["month"]=df_usa.job_posted_date.dt.month_name()
df_usa["month_number"]=df_usa.job_posted_date.dt.month
```

```python
df_usa_pivot=df_usa.pivot_table(index=["month","month_number"],values="salary_year_avg",columns="job_title_short",aggfunc="size")
df_usa_pivot=df_usa_pivot.sort_values("month_number")
df_usa_pivot = df_usa_pivot.droplevel("month_number", axis=0)
```

```python
df_usa_pivot
```

```python
a=df_usa.job_title_short.value_counts().head(3)
a=a.index.to_list()
a
df_usa_pivot=df_usa_pivot[a]
```

```python
df_usa_pivot.plot(kind="line")
```

```python
df_usa_pivot.index.name = "job_posted_month"
```
